```python
#import the necessary modules
import cv2
import numpy as np
import logging
```

```python
# we add logger with INFO level and  format that will track the events, you can also use print() function if u want to skip logger
logging.basicConfig(level=logging.INFO, format='[%(levelname)s] %(message)s')
```

```python
#make class to detects BGR color values from webcam feed and matches them to nearest color names
class ColorDetector:

    def __init__(self):
        # make Flags and give default values at initialization
        self.clicked = False
        self.bgr_color = (0, 0, 0)
        self.color_name = "Unknown"
        self.window_name = "Color Detector"

        # now, make Predefined color dictionary (BGR format mapped to simple color names), you can also add other
        self.color_dict = {
            (255, 0, 0): "Blue",
            (0, 255, 0): "Green",
            (0, 0, 255): "Red",
            (0, 255, 255): "Yellow",
            (255, 0, 255): "Magenta",
            (255, 255, 0): "Cyan",
            (0, 0, 0): "Black",
            (255, 255, 255): "White",
            (128, 128, 128): "Gray",
            (0, 128, 255): "Orange",
            (203, 192, 255): "Pink"
        }
        
     #make function to match the clicked BGR color to the closest predefined color.
     # Uses Euclidean distance in color space
    def get_closest_color(self, bgr):

        min_dist = float('inf')
        closest = "Unknown"
        for bgr_val, name in self.color_dict.items():
            dist = np.linalg.norm(np.array(bgr) - np.array(bgr_val))
            if dist < min_dist:
                min_dist = dist
                closest = name
        return closest
    
    # this function will triggered on mouse click and captures the BGR color at the clicked pixel.
    def mouse_callback(self, event, x, y, flags, frame):
        if event == cv2.EVENT_LBUTTONDOWN:
            self.bgr_color = frame[y, x].tolist()
            self.color_name = self.get_closest_color(self.bgr_color)
            self.clicked = True
            logging.info(f"Clicked at ({x}, {y}) - BGR: {self.bgr_color}, Name: {self.color_name}")

    # This will initializes webcam capture and checks for availability
    def setup_webcam(self):
        cap = cv2.VideoCapture(0)
        if not cap.isOpened():
            raise Exception("Webcam not accessible.")
        return cap

    # this is the main loop to capture webcam feed, detect color on click, and will display the result
    def run(self):
        try:
            cap = self.setup_webcam()
            cv2.namedWindow(self.window_name)
            logging.info("Webcam started. Click on the video to detect color. Press ESC to exit.")

            while True:
                ret, frame = cap.read()
                if not ret:
                    logging.error("Failed to grab frame.")
                    break

                # now, bind the mouse callback to the window
                cv2.setMouseCallback(self.window_name, self.mouse_callback, frame)

                # if  user clicked, rectangle will be formed to display color name
                if self.clicked:
                    text = f"BGR: {self.bgr_color} | Name: {self.color_name}"
                    brightness = sum(self.bgr_color)
                    text_color = (255, 255, 255) if brightness < 400 else (0, 0, 0)

                    # rectangle drawing 
                    cv2.rectangle(frame, (10, 10), (440, 50), self.bgr_color, -1)
                    cv2.putText(frame, text, (15, 40), cv2.FONT_HERSHEY_SIMPLEX,
                                0.7, text_color, 2, cv2.LINE_AA)

                # Displaying the frame
                cv2.imshow(self.window_name, frame)

                #windows will exit if ESC key is pressed
                if cv2.waitKey(1) & 0xFF == 27:
                    logging.info("ESC pressed. Exiting application.")
                    break

                # windows will also exit if the window is closed manually
                if cv2.getWindowProperty(self.window_name, cv2.WND_PROP_VISIBLE) < 1:
                    logging.info("Window closed. Exiting application.")
                    break

        except Exception as e:
            logging.exception(f"An unexpected error occurred: {e}")

        finally:
            # Release webcam and close all OpenCV windows
            cap.release()
            cv2.destroyAllWindows()
            logging.info("Webcam released and windows closed.")
```

```python
# now run the color detector
if __name__ == "__main__":
    detector = ColorDetector()
    detector.run()
```
